```python
import numpy as np
```

```python
Y = np.array([
    [5, 3, 0, 1],
    [0, 4, 5, 0],
    [3, 0, 2, 5]
])
print(Y)
```

```python
num_users, num_items = Y.shape
num_features = 2

X = np.random.randn(num_users, num_features) *1.000
W = np.random.randn(num_items, num_features) *1.000
print(W)
```

```python
prediction=np.dot(X,W.T)
squared_error= (prediction - Y) ** 2
```

```python
mask=(Y >0)
final_error=squared_error[mask]
print(final_error)
```

```python
cost=np.mean(final_error)
print(cost)
```

```python
epochs=1000
learning_rate = 0.01
reg_lambda = 0.0001
num_ratings = np.sum(mask)

for epoch in range (epochs):
    prediction=np.dot(X,W.T)
    error=(prediction-Y)*mask
    grad_x = (np.dot(error, W) + reg_lambda * X) / num_ratings
    grad_w = (np.dot(error.T, X) + reg_lambda * W) / num_ratings
    X = X - learning_rate * grad_x
    W = W - learning_rate * grad_w
    cost = np.sum(error**2) / np.sum(mask)

    if epoch % 100 == 0:
        print(f"Epoch {epoch} | Cost: {cost:.4f}")
```

```python
print("Original Ratings (Y)")
print(Y)

print("\nFinal Predictions")
final_preds = np.clip(np.dot(X, W.T), 1, 5)
print(np.round(final_preds, 2))
```
